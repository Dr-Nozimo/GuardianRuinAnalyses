```python
import numpy as np
import os
import matplotlib.pyplot as plt

# Folder with exported signals
folder_path = '../extracted_signals/'

# List all CSV files of the correct type
csv_files = [f for f in os.listdir(folder_path) if f.endswith('.csv') and f.startswith('EOL_')]

# Loop through the CSV files, load them, and add them to a list
extracted_arrays = []
for file in csv_files:
    file_path = os.path.join(folder_path, file)
    array = np.genfromtxt(file_path, delimiter=',')
    extracted_arrays.append(array)

# Convert the list of arrays into a NumPy array
extracted_arrays_np = np.array(extracted_arrays)

# Calculate the average across all arrays (along the first axis)
average_array = np.mean(extracted_arrays_np, axis=0)
```

```python
plt.imshow(average_array, origin='lower', cmap="inferno")
plt.colorbar()
```

```python
# Save the average array to a new CSV file
np.savetxt('../signal_samples/EOL_avg2_48k-2000-75-Taylor.csv', average_array, delimiter=',')
```
